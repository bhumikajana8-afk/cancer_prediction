```python
print("Survival Status Distribution:")

print(df["status"].value_counts())

sns.countplot(data=df, x="status", hue="status", legend=False)

plt.title("Breast Cancer Survival Status")
plt.xlabel("Status")
plt.ylabel("Number of Records")

plt.show()
```

```python
print("Missing Values:")

print(df.isnull().sum())

print("\nDuplicate Records:")

print(df.duplicated().sum())
```

```python
# Separate input features and target

X = df.drop("status", axis=1)

y = df["status"]

# Convert categorical columns into numerical columns

X = pd.get_dummies(X, drop_first=True, dtype=int)

print("Input Shape:", X.shape)

print("Target Shape:", y.shape)

display(X.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])

print("Testing records:", X_test.shape[0])
```

```python
model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)

print("Gradient Boosting Model Trained Successfully!")
```

```python
y_pred = model.predict(X_test)

print("Predicted Status:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy: {:.2f}%".format(accuracy * 100))
```

```python
print("Classification Report")

print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Breast Cancer Survival Prediction")
plt.xlabel("Predicted Status")
plt.ylabel("Actual Status")

plt.show()
```

```python
import joblib

joblib.dump(model, "cancer_gradient_boosting_model.pkl")

print("Model saved successfully!")
```
